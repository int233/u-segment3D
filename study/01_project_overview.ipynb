# u-Segment3D 学习笔记 01：项目概览

本笔记旨在快速建立对 u-Segment3D 项目的整体认知，包括论文思想、仓库结构以及环境配置建议。


## 1. 核心思想回顾
- **2D→3D 共识**：u-Segment3D 将多个正交视角的 2D 预测结果（概率图或实例 mask）进行聚合，通过距离变换、流场、梯度衰减等方式得到一致的 3D 实例分割。
- **Direct vs. Indirect**：Direct 方法需要概率+梯度；Indirect 方法只依赖实例 mask，适合已有 2D 分割模型但缺少概率图的场景。
- **模块化设计**：算法被拆分为参数管理、2D 预处理、三维聚合、后处理等子模块，方便针对不同数据形态调参。


## 2. 仓库结构速览
1. `segment3D/`：核心 Python 包，包含 `usegment3d.py`、`parameters.py`、`flows.py`、`watershed.py` 等。
2. `tutorials/`：官方教程脚本，涵盖单细胞、多细胞、组织等场景。
3. `example_data/`：用于快速体验的示例数据（需自行下载并解压）。
4. `docs/`：文档资源。
5. `PSFs/`：点扩散函数示例，用于各向异性数据的 PSF 补偿。
